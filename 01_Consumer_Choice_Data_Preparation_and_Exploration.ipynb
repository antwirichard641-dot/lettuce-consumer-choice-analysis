## Research Overview

This project investigates consumer acceptance of sustainably produced vegetables, focusing on hydroponic and traditional lettuce production and the role of organic labelling. It examines consumer perceptions, purchasing preferences, and willingness to pay using existing survey data. The aim is to understand how production methods and product attributes may influence consumer choices and the potential market acceptance of alternative agricultural production systems.

```python
# Upload and inspect the dataset structure
from google.colab import files
import pandas as pd

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name, low_memory=False)

print("Dataset dimensions:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Identify the variables directly related to consumer acceptance
key_variables = [
    "ID", "Willing_TradNonOrg", "Willing_TradOrg",
    "Willing_HydroNonOrg", "Willing_HydroOrg",
    "Organic", "Hydro", "Price", "Buy", "Choice"
]

print(df[key_variables].head())
```

### Relevant Research Outputs

The initial dataset inspection identified variables related to consumer willingness to purchase lettuce produced using traditional and hydroponic methods, with and without organic labelling. The dataset also contains purchase-choice variables, including price, buying decisions, and choice outcomes. These variables will support the investigation of how production methods and organic labelling relate to consumer acceptance and purchasing preferences.

```python
# Check how purchase and choice outcomes are coded
print(df[["Buy", "Choice", "NoBuy"]].value_counts(dropna=False).sort_index())
```

```python
# Check whether purchase and choice variables are consistent
print("Buy and Choice combinations:")
print(pd.crosstab(df["Buy"], df["Choice"], margins=True))

print("\nNoBuy values by Buy status:")
print(pd.crosstab(df["Buy"], df["NoBuy"], margins=True))

print("\nUnique values in each variable:")
for col in ["Buy", "Choice", "NoBuy"]:
    print(f"{col}: {sorted(df[col].dropna().unique())}")
```

```python
# Summarize purchase outcomes by production attributes and price
print(
    df.groupby(["Hydro", "Organic", "Price"], dropna=False)
      .agg(
          Observations=("Buy", "size"),
          Purchases=("Buy", "sum"),
          Purchase_Rate=("Buy", "mean")
      )
      .reset_index()
      .round(3)
      .to_string(index=False)
)
```

```python
# Inspect choice-set structure and outcome coding
cols = ["ID", "cset", "Alternative", "Hydro", "Organic",
        "Price", "Buy", "Choice", "NoBuy"]

print(df[cols].head(20).to_string(index=False))
```

```python
# Check whether each participant-choice set has exactly one selected option
choice_check = (
    df.groupby(["ID", "cset"])
      .agg(
          Alternatives=("Alternative", "size"),
          Selected=("Choice", "sum")
      )
)

print("Number of participant-choice sets:", len(choice_check))
print("\nAlternatives per choice set:")
print(choice_check["Alternatives"].value_counts().sort_index())
print("\nSelected alternatives per choice set:")
print(choice_check["Selected"].value_counts().sort_index())
```

```python
# Summarize selected alternatives by type
selected = df[df["Choice"] == 1].copy()

print("Selected alternatives:")
print(selected["Alternative"].value_counts().sort_index())

print("\nSelected alternative characteristics:")
print(
    selected.groupby("Alternative")
    .agg(
        Selections=("Choice", "size"),
        Hydroponic=("Hydro", "mean"),
        Organic=("Organic", "mean"),
        Mean_Price=("Price", "mean")
    )
    .round(3)
    .to_string()
)
```

The dataset contains 1,584 choice sets, each with two lettuce alternatives and one no-purchase alternative. Alternative 1 was selected 592 times (37.4%), Alternative 2 was selected 561 times (35.4%), and the no-purchase alternative was selected 431 times (27.2%). Overall, a lettuce option was selected in 72.8% of choice sets. These descriptive findings provide a baseline for subsequent analysis of the effects of hydroponic production, organic labelling, and price on consumer choices.

```python
# Check missing values in the main choice-model variables
model_variables = [
    "ID", "cset", "Alternative", "Choice",
    "Hydro", "Organic", "Price"
]

missing_summary = pd.DataFrame({
    "Missing_Count": df[model_variables].isna().sum(),
    "Missing_Percent": (df[model_variables].isna().mean() * 100).round(2)
})

print("Missing-value summary:")
print(missing_summary)

print("\nDataset dimensions:", df.shape)
print("Unique participants:", df["ID"].nunique())
print("Unique choice sets:", df[["ID", "cset"]].drop_duplicates().shape[0])
```

```python
# Save the validated choice-model dataset
model_variables = [
    "ID", "cset", "Alternative", "Choice",
    "Hydro", "Organic", "Price"
]

choice_df = df[model_variables].copy()
choice_df.to_csv("lettuce_consumer_choice_clean.csv", index=False)

print("Saved: lettuce_consumer_choice_clean.csv")
print("Dataset dimensions:", choice_df.shape)
```
